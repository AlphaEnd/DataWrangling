```python
import pandas as pd
import numpy as np

# Atur biar angka desimalnya hanya 2 digit
pd.set_option('display.float_format', lambda x: '%.2f' % x)
# Siapkan data nya
df = pd.read_csv(r"B:\Python\DataSet\CSV\Zomato Restaurant Dataset.csv")
'''
['Restaurant ID',
 'Restaurant Name',
 'Country Code',
 'City',
 'Address',
 'Locality',
 'Locality Verbose',
 'Longitude',
 'Latitude',
 'Cuisines',
 'Average Cost for two',
 'Currency',
 'Has Table booking',
 'Has Online delivery',
 'Is delivering now',
 'Switch to order menu',
 'Price range',
 'Aggregate rating',
 'Rating color',
 'Rating text',
 'Votes']
'''
```

```python
# Filter dengan satu kondisi
df[df['Aggregate rating'] < 2.5]
```

```python
# Filter dengan dua kondisi (& = AND)
df_filtered = df[
    (df["Aggregate rating"] > 4.0) &
    (df["City"] == 'Jakarta')
]

df_filtered
```

```python
# Mengurutkan data
df_sorted = df.sort_values(
    by="City",
    ascending=True
)
df_sorted
'''
1. by                : menentukan kolom mana yang digunakan untuk mengurutkan
2. ascending=False   : berarti urutan menurun, dari yang terbesar ke yang terkecil
3. ascending=True    : berarti urutan naik, dari terkecil ke yang terbesar
'''
```

```python
# Filter dengan dua kondisi (| = OR)
df[
    (df['City'] == 'Jakarta') |
    (df['Aggregate rating'] > 4.3)
]
```

1. < atau >         : Membandingkan nilai
2. ==               : Memeriksa kesamaan
3. &                : Kedua kondisi harus benar
4. \|               : Minimal satu kondisi benar
5. sort_values()    : Mengurutkan baris berdasarkan kolom
6. ascending=False  : Urutan menurun

Pada penggunaan kondisi lebih dari satu, maka setiap kondisi harus digapit oleh tanda kurung () dan dihubungkan dnegan operator logika seperti '&' untuk AND dan '\|' untuk OR.
