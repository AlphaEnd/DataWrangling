```python
# Menghitung nilai kosong
df.isnull().sum()
```

```python
# Menghapus nilai kosong
df_cleaned = df.dropna()
```

```python
# Mengisi data kosong dengan data default
df_copy = df.copy() # Copy dulu biar melek
df_copy['Cuisines'] = df_copy['Cuisines'].fillna("Unknown")

# Mengisi nilai kosong dnegan nilai hasil agregate
# df_copy['Aggregate rating'] = df_copy['Aggregate rating'].fillna(df_copy['Aggregate rating'].mean())

print(df_copy[df_copy['Cuisines'] == 'Unknown'][['Cuisines', 'City']])
```

```python
# Memperkirakan nilai kosong dengan interpolasi (Berguna untuk data berurutan)

angka = pd.Series([10, 20, 30, np.nan, 64, 128, 256])
# 30 + (64 - 30) / 2
angka.interpolate()
```

1. isnull().sum() => Menghitung nilai kosong => Langkah pemeriksaan awal
2. dropna() => Menghapus baris kosong => Data yang tersisa bisa berkurang
3. fillna() => Mengganti nilai kosong => Nilai pengganti harus masuk akal
4. interpolate() => Memperkirakan nilai di antara data => Cocok jika data mempunyai urutan
