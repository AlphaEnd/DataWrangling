```python
# Penggunaan marge() => Mencocokan kunci
info = pd.DataFrame({
    "Restaurant ID": [101, 102, 103, 104],
    "Nama": ["Cafe A", "Cafe B", "Cafe C", "Cafe D"]
})

rating = pd.DataFrame({
    "Restaurant ID": [101, 102, 103, 105],
    "Rating": [4.5, 3.8, 4.2, 4.0]
})

hasil = info.merge(rating, on="Restaurant ID", how="outer")
print(hasil)
```

| Parameter | Fungsi |
|---|---|
| `on=` | Kolom kunci untuk mencocokkan data |
| `how="inner"` | Hanya mempertahankan kunci yang cocok di kedua tabel |
| `how="left"` | Mempertahankan seluruh baris dari tabel kiri |
| `how="outer"` | Mempertahankan semua kunci dari kedua tabel |

```python
# Penggunaan concat() => Manyambungkan tabel

jakarta = pd.DataFrame({
    "Nama": ["Cafe A"],
    "Rating": [4.5]
})

bandung = pd.DataFrame({
    "Nama": ["Cafe B"],
    "Rating": [4.2]
})

gabungan = pd.concat([jakarta, bandung], ignore_index=True) # ignore_index=True membuat index hasil menjadi berurutan dari 0
display(gabungan)
```

```python
# Contoh lain
info = df[["Restaurant ID", "Restaurant Name"]]
rating = df[["Restaurant ID", "Aggregate rating"]]

hasil = info.merge(
    rating,
    on="Restaurant ID",
    how="inner",
    validate="one_to_one" # Memastikan bahwa setiap ID hanya muncul satu kali
)

hasil.head()
```

| Kebutuhan | Metode |
|---|---|
| Menggabungkan tabel berdasarkan ID | `merge()` |
| Mempertahankan semua baris dari tabel kiri | `merge(how="left")` |
| Mempertahankan kunci dari kedua tabel | `merge(how="outer")` |
| Menyambungkan baris dari beberapa tabel | `concat()` |
| Memeriksa asumsi keunikan kunci | `validate=` pada `merge()` |
