```python
# Penanganan data besar dengan chunksize

total_restoran = 0

for chunk in pd.read_csv(
    "data_besar.csv",
    chunksize=100000 # Membaca data dalam potongan hingga 100.000 baris
):
    total_restoran += len(chunk)

print(total_restoran)
```

```python
# Mengunakan kolom yang benar-benar diperlukan

df_kecil = pd.read_csv(
    r"B:\Python\DataSet\CSV\Zomato Restaurant Dataset.csv",
    usecols=["City", "Aggregate rating"]
)

# Memeriksa penggunaan memori
df_kecil.info(memory_usage="deep")

df.info(memory_usage="deep")
df.memory_usage(deep=True) # Memeriksa penggunaan memori per kolom
```

```python
# Menurunkan skala tipe data numerik (Downcasting)
# Manual
df["Price range"] = df["Price range"].astype("int8")
df["Aggregate rating"] = df["Aggregate rating"].astype("float16")

# Otomatis: Memilih tipe terkecil yang cocok
df["Average Cost for two"] = pd.to_numeric(df["Average Cost for two"], downcast="integer")

# Gunakan category untuk text berulang
df["City"] = df["City"].astype("category")
```
