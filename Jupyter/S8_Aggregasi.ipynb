```python
import pandas as pd
import numpy as np

# Buat data mini
df1 = pd.DataFrame({
    'Restaurant': ['Cafe A', 'Cafe B', 'Cafe C'],
    'Rating': [4.5, 3.2, 3.9],
    'Value': [2, 4, 5]
})

# Atur biar angka desimalnya hanya 2 digit
pd.set_option('display.float_format', lambda x: '%.2f' % x)
# Siapkan data nya
df = pd.read_csv(r"B:\Python\DataSet\CSV\Zomato Restaurant Dataset.csv")
'''
['Restaurant ID',
 'Restaurant Name',
 'Country Code',
 'City',
 'Address',
 'Locality',
 'Locality Verbose',
 'Longitude',
 'Latitude',
 'Cuisines',
 'Average Cost for two',
 'Currency',
 'Has Table booking',
 'Has Online delivery',
 'Is delivering now',
 'Switch to order menu',
 'Price range',
 'Aggregate rating',
 'Rating color',
 'Rating text',
 'Votes']
'''

df.head()
```

```python
# Menghitung rata-rata per kota
df.groupby("City")["Aggregate rating"].mean()

# groupby("City") => Mengelompokan brais berdasarkan kota
# ["Aggregate rating"] => Memilih kolom yang ingin dihitung
# .mean() => Menghitung rata-rata rating dari setiap kelompok
```

```python
# Menghiutng jumlah restoran
df.groupby("City")["Restaurant Name"].count() # Menghitung jumlah baris pada setiap kelompok yang tidak kosong

# Menghitung baris pada setiap kelompok
df.groupby("City").size()
```

```python
# Menghitung beberapa agregasi sekaligus
df.groupby("City")["Aggregate rating"].agg(
    ["mean", "min", "max", "count", "median"]
).sort_values(ascending=False, by=["count", "mean"]).head(10)
```

```python
# Agregasi multi index
cara1 = df.groupby("City").agg(
    total_restaurant = ("Restaurant Name", "count"),
    average_rating = ("Aggregate rating", "mean"),
    std = ("Aggregate rating", "std"),
    avg_price = ("Average Cost for two", "mean")
).reset_index()

# Sintaks : output_column_name = (input_column_name, aggregation_function)
# Bentuk ke dua
cara2 = df.groupby("City").agg(**{
    "Total Restaurant" : ("Restaurant Name", "count"),
    "Average Rating" : ("Aggregate rating", "mean"),
    "Standard Deviation" : ("Aggregate rating", "std"),
    "Average Price for two" : ("Average Cost for two", "mean")
}).reset_index()

print(cara2)
```

```python
# Perhitungan setelah agregasi
hitungA = (df.groupby("City").agg(**{
    "Total Restaurant" : ("Restaurant Name", "count"),
    "Average Rating" : ("Aggregate rating", "mean"),
    "Standard Deviation" : ("Aggregate rating", "std"),
    "Average Price for two" : ("Average Cost for two", "mean")
})
.assign(
    price_for_one = lambda x: x["Average Price for two"] / 2,
    twice_restaurant = lambda x: x["Total Restaurant"] * 2
)
.reset_index()
)

display(hitungA)
```

1. groupby() => Membagi data menjadi kelompok berdasarkan satu atau lebih kolom.
2. mean() => Menghitung rata-rata.
3. count() => Menghitung jumlah nilai yang tidak kosong pada kolom yang dipilih.
4. size() => Menghitung jumlah baris pada setiap kelompok.
5. agg() => Menerapkan satu atau beberapa fungsi agregasi.
