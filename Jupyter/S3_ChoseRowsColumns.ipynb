```python
df1 = pd.DataFrame(
    {"Index": [11, 22, 33],
     "Name": ['Cafe A', 'Cafe B', 'Cafe C'],
     "City": ['New York', 'Los Angeles', 'Chicago'],
     "Rating": [4.5, 4.0, 3.5]}
)
df1 = df1.set_index("Index")

# Mengambil baris dengan index 11
print(df1.loc[11]) # .loc[baris, nama kolom] atau .loc[baris] untuk semua kolom
print("\n")
# Mengambil baris ke-n (harus angka coy)
print(df1.iloc[0]) # .iloc[baris, kolom] atau .iloc[baris] untuk semua kolom
# Slicing rows
print("\n", df1.iloc[1:]) # Hampir sama dengan indexing python pada umumnya
```
