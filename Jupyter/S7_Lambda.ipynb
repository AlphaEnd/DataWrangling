```python
import pandas as pd
import numpy as np

# Buat data mini
df1 = pd.DataFrame({
    'Restaurant': ['Cafe A', 'Cafe B', 'Cafe C'],
    'Rating': [4.5, 3.2, 3.9],
    'Value': [2, 4, 5]
})

# Atur biar angka desimalnya hanya 2 digit
pd.set_option('display.float_format', lambda x: '%.2f' % x)
# Siapkan data nya
df = pd.read_csv(r"B:\Python\DataSet\CSV\Zomato Restaurant Dataset.csv")
'''
['Restaurant ID',
 'Restaurant Name',
 'Country Code',
 'City',
 'Address',
 'Locality',
 'Locality Verbose',
 'Longitude',
 'Latitude',
 'Cuisines',
 'Average Cost for two',
 'Currency',
 'Has Table booking',
 'Has Online delivery',
 'Is delivering now',
 'Switch to order menu',
 'Price range',
 'Aggregate rating',
 'Rating color',
 'Rating text',
 'Votes']
'''
```

```python
# Membuat fungsi
def kategori_rating(rating):
    if rating >= 4.0:
        return 'Tinggi'
    else:
        return 'Rendah'
```

```python
# Menerpakan fungsi ke seluruh kolom
df1['Kategori'] = df1['Rating'].apply(kategori_rating)
# df1['Kategori'] membuat kolom baru
df1
```

```python
# Versi lambda (Biasanya untuk fungsi pendek)
df1['Kategori'] = df1['Rating'].apply(
    lambda rating: 'Tinggi' if rating >= 4.0 else 'Rendah'
)

df1
```

1. lambda rating: => Membuat fungsi kecil dengan masukan bernama rating
2. rating >= 4.0 => memeriksa kondisi
3. "Tinggi" if ... else "Rendah" => memilih hasil berdasarkan kondisi
4. .apply(...) => menerapkan fungsi pada setiap nilai di kolom

NB: Untuk fungsi sederhana, lambda uckup praktis. Untuk fungsi ynag panjang atau kompleks, fungsi def biasanya lebih mudah dibaca
