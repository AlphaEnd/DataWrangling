```python
import pandas as pd
import numpy as np

# Buat data mini
df1 = pd.DataFrame({
    'Restaurant': ['Cafe A', 'Cafe B', 'Cafe C'],
    'Rating': [4.5, 3.2, 3.9],
    'Value': [2, 4, 5]
})

# Atur biar angka desimalnya hanya 2 digit
pd.set_option('display.float_format', lambda x: '%.2f' % x)
# Siapkan data nya
df = pd.read_csv(r"B:\Python\DataSet\CSV\Zomato Restaurant Dataset.csv")
'''
['Restaurant ID',
 'Restaurant Name',
 'Country Code',
 'City',
 'Address',
 'Locality',
 'Locality Verbose',
 'Longitude',
 'Latitude',
 'Cuisines',
 'Average Cost for two',
 'Currency',
 'Has Table booking',
 'Has Online delivery',
 'Is delivering now',
 'Switch to order menu',
 'Price range',
 'Aggregate rating',
 'Rating color',
 'Rating text',
 'Votes']
'''

df.head()
```

```python
# Menggunakan data contoh pada pivot table

data = pd.DataFrame({
    "Kota": ["Jakarta", "Jakarta", "Bandung", "Bandung", "Jakarta"],
    "Jenis": ["Cafe", "Restoran", "Cafe", "Restoran", "Cafe"],
    "Rating": [4.5, 4.0, 4.2, 3.8, 3.7]
})

tabel = data.pivot_table(
    index="Kota",
    columns="Jenis",
    values="Rating",
    aggfunc="mean"
)

tabel
```

> `pivot_table()` => Mengubah data menjadi ringkasan berbentuk tabel berdasarkan kategori

| Parameter | Fungsi |
|---|---|
| `index="Kota"` | kota menjadi baris |
| `columns="Jenis"` | jenis restoran menjadi kolom |
| `values="Rating"` | nilai yang diringkas |
| `aggfunc="mean"` | hitung rata-rata | 

```python
# melt() - mnegubah kolom menjadi baris

data_panjang = tabel.reset_index().melt(
    id_vars="Kota",
    var_name="Jenis",
    value_name="Rating"
)

data_panjang

# melt() membantu mengubah data yang memiliki banyak kolom menjadi format yang lebih mudah dianalisis
```

```python
# Cotoh pada DataSet Zomato
# Menggunakan pivot_table()

ringkasan = df.pivot_table(
    index="City",
    columns="Price range",
    values="Aggregate rating",
    aggfunc="mean"
)

ringkasan.head()
```

```python
# Menggunakan melt()

ringkasan_panjang = ringkasan.reset_index().melt(
    id_vars="City",
    var_name="Price range",
    value_name="Average rating"
)

ringkasan_panjang.head()
```

| Kebutuhan | Metode |
|---|---|
| Membuat tabel ringkasan berdasarkan kategori | `pivot_table()` |
| Mengubah kolom menjadi baris | `melt()` |
| Menentukan kelompok baris | `index=` |
| Menentukan kelompok kolom | `columns=` |
| Menentukan nilai yang dihitung | `values=` |
| Menentukan perhitungan | `aggfunc=` |
