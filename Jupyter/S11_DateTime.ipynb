```python
# Buat contoh data
df_waktu = pd.DataFrame({
    "Tanggal": ["2026-01-05", "2026-01-12", "2026-02-03", "2026-02-11", "2026-01-25"],
    "Penjualan": [100, 150, 200, 120, 100]
})

# Mengubah teks menjadi tanggal
df_waktu["Tanggal"] = pd.to_datetime(df_waktu["Tanggal"])

print(df_waktu.dtypes)
```

```python
# Mengambil tahun dan bulan

df_waktu["Tahun"] = df_waktu["Tanggal"].dt.year
df_waktu["Bulan"] = df_waktu["Tanggal"].dt.month_name()

df_waktu
```

```python
# Menghitung total perbulan

df_waktu["Bulan"] = df_waktu["Tanggal"].dt.to_period("M")

total_bulanan = df_waktu.groupby("Bulan")["Penjualan"].sum()

print(total_bulanan)

'''
groupby() mengelompokan data berdasarkan bulan, 
lalu sum() menjumlahkan penjualan setiap kelompok  
'''
```

```python
'''
resample() digunakan untuk mengagregasi data berdasarkan interval waktu.
Data harus memiliki DatetimeIndex atau menggunakan kolom tanggal dengan
mekanisme yang sesuai.
'''

df_resample = df_waktu.set_index("Tanggal")

total_bulanan = df_resample["Penjualan"].resample("MS").sum()

print(total_bulanan)

'''
"MS" berarti awal setiap bulan (month start). Perlu diingat bahwa hasil resample()
dapat mencakup interval waktu yang tidak memiliki transaksi; hasil penjumlahannyan bisa nol.
'''
```

| Kebutuhan | Metode |
|---|---|
| Mengonversi teks menjadi tanggal | `pd.to_datetime()` |
| Mengambil tahun | `.dt.year` |
| Mengambil bulan | `.dt.month` |
| Mengambil nama bulan | `.dt.month_name()` |
| Mengelompokkan berdasarkan bulan | `groupby()` atau `resample()` |
| Mengatur tanggal sebagai indeks | `set_index()` |
