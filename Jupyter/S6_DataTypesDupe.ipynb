```python
import pandas as pd
import numpy as np

# Buat data mini
df1 = pd.DataFrame({
    'Restaurant': ['Cafe A', 'Cafe B', 'Cafe A'],
    'Rating': ['4.2', '3.8', '4.2'],
    'Votes': [120, 80, 120]
})

# Atur biar angka desimalnya hanya 2 digit
pd.set_option('display.float_format', lambda x: '%.2f' % x)
# Siapkan data nya
df = pd.read_csv(r"B:\Python\DataSet\CSV\Zomato Restaurant Dataset.csv")
'''
['Restaurant ID',
 'Restaurant Name',
 'Country Code',
 'City',
 'Address',
 'Locality',
 'Locality Verbose',
 'Longitude',
 'Latitude',
 'Cuisines',
 'Average Cost for two',
 'Currency',
 'Has Table booking',
 'Has Online delivery',
 'Is delivering now',
 'Switch to order menu',
 'Price range',
 'Aggregate rating',
 'Rating color',
 'Rating text',
 'Votes']
'''
```

```python
# Memeriksa tipe data
df1.dtypes
```

```python
# Mengubah tipe data dengan astype()
df1['Rating'] = df1['Rating'].astype(float)

df1.dtypes
```

```python
# Menghitung rata-rata setelah mengubah tipe data
df1['Rating'].mean()
```

```python
# Menampilkan baris yang merupakan duplikat
df1[df1.duplicated()]

# Buat DataFrame baru tanpa duplikat
df_unique = df1.drop_duplicates()

df_unique
```

Secara default, duplicated() menandai baris yang sama dengan baris sebelumnya, dan drop_duplicates() mempertahankan kemunculan pertama. Dalam contoh ini, satu baris Cafe A dihapus dari hasil.

```python
# Penerapan pada DataSet Zomato

# Periksa tipe data setiap kolom
df.dtypes

# Hitung jumlah data duplikat
df.duplicated().sum()

# Buat salinan tanpa baris yang identik
df_unq = df.drop_duplicates()

# Membandingkan jumlah baris sebelum dan sesudah
print("Sebelum:", len(df))
print("Sesudah:", len(df_unq))
```

1. df.dtypes            : Memeriksa tipe data kolom
2. astype(float)	    : Mengubah nilai menjadi angka desimal
3. duplicated().sum()	: Menghitung baris duplikat
4. drop_duplicates()	: Menghapus baris duplikat
5. pd.to_datetime()	    : Mengonversi nilai ke tipe tanggal/waktu
