```python
import pandas as pd
import numpy as np

# Buat data mini
df1 = pd.DataFrame({
    'Restaurant': ['Cafe A', 'Cafe B', 'Cafe C'],
    'Rating': [4.5, 3.2, 3.9],
    'Value': [2, 4, 5]
})

# Atur biar angka desimalnya hanya 2 digit
pd.set_option('display.float_format', lambda x: '%.2f' % x)
# Siapkan data nya
df = pd.read_csv(r"B:\Python\DataSet\CSV\Zomato Restaurant Dataset.csv")
'''
['Restaurant ID',
 'Restaurant Name',
 'Country Code',
 'City',
 'Address',
 'Locality',
 'Locality Verbose',
 'Longitude',
 'Latitude',
 'Cuisines',
 'Average Cost for two',
 'Currency',
 'Has Table booking',
 'Has Online delivery',
 'Is delivering now',
 'Switch to order menu',
 'Price range',
 'Aggregate rating',
 'Rating color',
 'Rating text',
 'Votes']
'''

df.head()
```

```python
# Ringkasan statistik dengan describe()
df["Aggregate rating"].describe()
```

>Hasilnya mencakup:

| Statistik | Makna |
|---|---|
| `count` | Jumlah nilai yang tidak kosong |
| `mean` | Rata-rata |
| `std` | Simpangan baku |
| `min` | Nilai minimum |
| `25%` | Kuartil pertama (Q1) |
| `50%` | Median (Q2) |
| `75%` | Kuartil ketiga (Q3) |
| `max` | Nilai maksimum |

```python
# Menghitung statistik secara terpisah
rating = df["Aggregate rating"]

print("Mean:", rating.mean())
print("Median:", rating.median())
print("Modus:", rating.mode().to_list())
print("Standard deviasi:", rating.std())
```

```python
# Bandingkan rating antar kota

ringkasan = (
    df.groupby("City")["Aggregate rating"]
    .agg(["count", "mean", "median", "std"])
    .sort_values("count", ascending=False)
)

ringkasan.head(10)
```

```python
# Periksa distribusi dan nilai ekstrem

rating.quantile([0.25, 0.50, 0.75, 0.95])
```

```python
# Memriksa rentang antar Quartil (Inter Quartil Range)

q1 = rating.quantile(0.25)
q3 = rating.quantile(0.75)

iqr = round(q3 - q1, 2)

batas_bawah = q1 - 1.5 * iqr
batas_atas = q3 + 1.5 * iqr

outlier = rating[
    (rating < batas_bawah) |
    (rating > batas_atas)
]

print("Jumlah kandidat outlier:", len(outlier))
```

| Tujuan | Metode |
|---|---|
| Ringkasan statistik | `.describe()` |
| Rata-rata | `.mean()` |
| Nilai tengah | `.median()` |
| Modus | `.mode()` |
| Penyebaran | `.std()` |
| Kuantil | `.quantile()` |
| Statistik per kelompok | `.groupby().agg()` |
| Kandidat outlier | Metode IQR |
