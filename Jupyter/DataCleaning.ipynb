```python
# 1. Memeriksa jumlah seluruh baris duplikat
print("Duplikat baris:", df.duplicated().sum())

# 2. Memeriksa Restaurant ID yang berulang
print("ID duplikat:", df["Restaurant ID"].duplicated().sum())

# 3. Melihat baris yang terindikasi duplikat
duplicates = df[df.duplicated(keep=False)]

display(duplicates.head(10))
```

```python
# 1. Memeriksa jumlah seluruh baris duplikat
print("Duplikat baris:", df.duplicated().sum())

# 2. Memeriksa Restaurant ID yang berulang
print("ID duplikat:", df["Restaurant ID"].duplicated().sum())

# 3. Melihat baris yang terindikasi duplikat
duplicates = df[df.duplicated(keep=False)]

display(duplicates.head(10))
```

1. df.duplicated()      : Menghasilkan True jika sebuah baris terdeteksi sebagai duplikat dari baris sebelumnya.
2. .sum()  	            : Menghitung jumlah nilai True, karena dalam Python True bernilai 1.
3. df["Restaurant ID"]  : Memilih satu kolom berdasarkan namanya.
4. keep=False           : Menandai semua anggota kelompok duplikat, bukan hanya kemunculan berikutnya.

```python
# A. Periksa rentang rating
invalid_rating = df[
    (df["Aggregate rating"] < 0) |
    (df["Aggregate rating"] > 5)
]

print("Rating di luar rentang:", len(invalid_rating))


# B. Periksa biaya negatif
invalid_cost = df[df["Average Cost for two"] < 0]

print("Biaya negatif:", len(invalid_cost))


# C. Periksa koordinat geografis
invalid_latitude = df[
    (df["Latitude"] < -90) |
    (df["Latitude"] > 90)
]

invalid_longitude = df[
    (df["Longitude"] < -180) |
    (df["Longitude"] > 180)
]

print("Latitude tidak valid:", len(invalid_latitude))
print("Longitude tidak valid:", len(invalid_longitude))
```

1. | berarti OR dalam operasi logika Pandas.
2. & berarti AND.
3. Setiap kondisi harus dibungkus dengan tanda kurung.
4. len() menghitung jumlah elemen dalam objek, dalam kasus ini jumlah baris yang terdeteksi.
